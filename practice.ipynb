```python
#  Try-Except

try:
    num = int(input("Enter a number: "))
    print(10 / num)

except:
    print("An error occurred")
```

```python
# Concept: Multiple Exceptions

try:
    num = int(input("Enter number: "))
    result = 10 / num

except ValueError:
    print("Enter an integer")

except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
#  Recursive Function

def countdown(n):
    if n == 0:       # Base Case
        return

    print(n)
    countdown(n - 1)  # Recursive Call

countdown(5)
```

```python
#  Factorial Using Recursion

def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```

```python
#  Sum Using Recursion

def total(n):
    if n == 0:
        return 0

    return n + total(n - 1)

print(total(5))
```
