```python
# Task: Calculate student grade using marks

marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
elif marks >= 60:
    print("Grade D")
else:
    print("Fail")
```

```python
# Task: Calculate electricity bill based on units

units = int(input("Enter consumed units: "))

if units <= 100:
    bill = units * 10
elif units <= 200:
    bill = 100 * 10 + (units - 100) * 15
else:
    bill = 100 * 10 + 100 * 15 + (units - 200) * 20

print("Total Bill:", bill)
```

```python
# Simple Calculator

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

op = input("Enter operator (+, -, *, /): ")

if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print(a * b)
elif op == "/":
    if b != 0:
        print(a / b)
    else:
        print("Cannot divide by zero")
else:
    print("Invalid Operator")
```

```python
# Task: Count even and odd numbers in a range

n = int(input("Enter range: "))

even = 0
odd = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)
```

```python
# Task: Calculate factorial using loop

n = int(input("Enter number: "))

fact = 1

for i in range(1, n + 1):
    fact *= i

print("Factorial:", fact)
```

```python
# Task: Reverse digits of a number

num = int(input("Enter number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed Number:", reverse)
```

```python
# Task: Calculate average using function

def average(a, b, c):
    return (a + b + c) / 3

x = int(input("Enter first number: "))
y = int(input("Enter second number: "))
z = int(input("Enter third number: "))

print("Average:", average(x, y, z))
```

```python
# Task: Check whether a number is prime

def check_prime(n):

    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


num = int(input("Enter number: "))

if check_prime(num):
    print("Prime Number")
else:
    print("Not Prime")
```

```python
# Task: Count how many times each element appears

numbers = [1, 2, 2, 3, 3, 3, 4, 4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
```
