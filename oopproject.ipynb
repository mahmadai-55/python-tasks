```python
# ATM System
class ATM:
    def __init__(self, balance=0):
        self.pin = ""
        self.balance = balance
        self.choose_action()


    def choose_action(self):
        while True:
            
            print("\n1. Enter PIN")
            print("2. Check Balance")
            print("3. Deposit")
            print("4. Withdraw")
            print("5. Exit")
            choice = input("Choose an option: ")
            if choice == "1":
                self.enter_pin()
            elif choice == "2":
                self.check_balance()
            elif choice == "3":
                self.deposit()
            elif choice == "4":
                self.withdraw()
            elif choice == "5":
                print("Thank you for using the ATM.")
                break
            else:
                print("Invalid option. Please try again.")

    def deposit(self):
        check = input("Enter your PIN to confirm deposit: ")
        if check == self.pin:
            amount = float(input("Enter deposit amount: "))
            if amount > 0:
                self.balance += amount
                print(f"Deposited: ${amount}. New balance: ${self.balance}.")
            else:
                print("Deposit amount must be positive.")
        else:
            print("Invalid PIN.")

    def withdraw(self):
        check = input("Enter your PIN to confirm withdrawal: ")
        if check == self.pin:
            amount = float(input("Enter withdrawal amount: "))
            if amount > 0:
                if amount <= self.balance:
                    self.balance -= amount
                    print(f"Withdrew: ${amount}. New balance: ${self.balance}.")
                else:
                    print("Insufficient funds.")
            else:
                print("Withdrawal amount must be positive.")
        else:
            print("Invalid PIN.")

    def check_balance(self):
        check = input("Enter your PIN to confirm balance inquiry: ")
        if check == self.pin:
            print(f"Current balance: ${self.balance}.")
        else:
            print("Invalid PIN.")

    def enter_pin(self):
        self.pin = input("Enter your PIN: ")
        print("PIN entered successfully.")

obj=ATM()
```

```python
# Creating a new data type
class Fraction:
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator

    def __str__(self):
        return f"{self.numerator}/{self.denominator}"

    def __add__(self, other):
        new_numerator = (self.numerator * other.denominator) + (other.numerator * self.denominator)
        new_denominator = self.denominator * other.denominator
        return Fraction(new_numerator, new_denominator)

    def __sub__(self, other):
        new_numerator = (self.numerator * other.denominator) - (other.numerator * self.denominator)
        new_denominator = self.denominator * other.denominator
        return Fraction(new_numerator, new_denominator)

    def __mul__(self, other):
        new_numerator = self.numerator * other.numerator
        new_denominator = self.denominator * other.denominator
        return Fraction(new_numerator, new_denominator)

    def __truediv__(self, other):
        new_numerator = self.numerator * other.denominator
        new_denominator = self.denominator * other.numerator
        return Fraction(new_numerator, new_denominator)

obj1 = Fraction(3, 2)
print("Fraction 1:", obj1)
obj2 = Fraction(5, 3)
print("Fraction 2:", obj2)
sum=obj1+obj2
print("Sum of fractions:", sum)
diff=obj1-obj2
print("Difference of fractions:", diff)
prod=obj1*obj2
print("Product of fractions:", prod)
quot=obj1/obj2
print("Quotient of fractions:", quot)
```
