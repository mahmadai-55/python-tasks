```python
#Task 1: Calculator Class
class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def subtract(self):
        return self.num1 - self.num2

    def multiply(self):
        return self.num1 * self.num2

    def divide(self):
        if self.num2 != 0:
            return self.num1 / self.num2
        return "Cannot divide by zero"

# Test object creation and operations
calc = Calculator(10, 5)
print("Addition:", calc.add())
print("Subtraction:", calc.subtract())
print("Multiplication:", calc.multiply())
print("Division:", calc.divide())
```

```python
#Task 5: Book Class (UML Diagram)
class Book:
    def __init__(self, title: str, author: str, price: float, copies: int):
        self.title = title
        self.author = author
        self.price = price
        self.copies = copies

    def display_info(self):
        print(f"Title: {self.title}, Author: {self.author}, Price: ${self.price}, Copies: {self.copies}")

    def sell(self, qty: int):
        if qty <= self.copies:
            self.copies -= qty
            print(f"Sold {qty} copy/copies.")
        else:
            print("Not enough copies in stock!")

    def restock(self, qty: int):
        self.copies += qty
        print(f"Restocked {qty} copy/copies.")

# Test
b1 = Book("Python Basics", "John Doe", 29.99, 10)
b1.display_info()
b1.sell(3)
b1.display_info()
b1.restock(5)
b1.display_info()
```

```python
#Task 6: Temperature Conversion Class (UML Diagram
class Temperature:
    def __init__(self, celsius: float):
        self.celsius = celsius

    def to_fahrenheit(self) -> float:
        return self.celsius * (9 / 5) + 32

    def to_kelvin(self) -> float:
        return self.celsius + 273.15

    def display(self):
        print(f"Celsius: {self.celsius}°C")
        print(f"Fahrenheit: {self.to_fahrenheit()}°F")
        print(f"Kelvin: {self.to_kelvin()}K")

# Test
temp = Temperature(25.0)
temp.display()
```

```python
#Task 7: MovieShow Class (UML Diagram
class MovieShow:
    def __init__(self, movie_name: str, available_seats: int, ticket_price: float):
        self.movie_name = movie_name
        self.available_seats = available_seats
        self.ticket_price = ticket_price

    def book(self, n: int):
        if n <= self.available_seats:
            self.available_seats -= n
            print(f"Successfully booked {n} seats.")
        else:
            print("Seats not available")

    def cancel(self, n: int):
        self.available_seats += n
        print(f"Cancelled {n} seats.")

    def show_details(self):
        print(f"Movie: {self.movie_name} | Available Seats: {self.available_seats} | Ticket Price: Rs. {self.ticket_price}")

# Test
show = MovieShow("Inception", 50, 500.0)
show.show_details()

show.book(5)
show.show_details()

show.book(50)  # Should fail

show.cancel(2)
show.show_details()
```

```python
#Task 8: Student Result
class StudentResult:
    def __init__(self, student_name, roll_number, m1, m2, m3):
        self.student_name = student_name
        self.roll_number = roll_number
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def calculate_total(self):
        return self.m1 + self.m2 + self.m3

    def calculate_percentage(self):
        return (self.calculate_total() / 300) * 100

    def display(self):
        print(f"Name: {self.student_name} | Roll No: {self.roll_number}")
        print(f"Total Marks: {self.calculate_total()}/300")
        print(f"Percentage: {self.calculate_percentage():.2f}%")

# Test
res = StudentResult("Ahmed", "BS-102", 85, 90, 78)
res.display()
```

```python
#Task 9: Library Book
class LibraryBook:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_issued = False

    def issue_book(self):
        if not self.is_issued:
            self.is_issued = True
            print(f"Book '{self.title}' has been issued successfully.")
        else:
            print("Book already issued")

    def return_book(self):
        if self.is_issued:
            self.is_issued = False
            print(f"Book '{self.title}' has been returned.")
        else:
            print("Book was not issued.")

    def display_details(self):
        status = "Issued" if self.is_issued else "Available"
        print(f"Title: {self.title} | Author: {self.author} | Status: {status}")

# Test
b1 = LibraryBook("Clean Code", "Robert C. Martin")
b2 = LibraryBook("1984", "George Orwell")

b1.issue_book()
b1.issue_book()  # Testing already issued condition
b1.display_details()
b1.return_book()
b1.display_details()
```

```python
#Task 10: Mobile SIM
class MobileSIM:
    def __init__(self, phone_number, owner_name, balance=0.0):
        self.phone_number = phone_number
        self.owner_name = owner_name
        self.balance = balance

    def recharge(self, amount):
        self.balance += amount
        print(f"Recharged Rs. {amount}. New Balance: Rs. {self.balance}")

    def make_call(self, minutes):
        cost = minutes * 2
        if self.balance >= cost:
            self.balance -= cost
            print(f"Call made for {minutes} min(s). Cost: Rs. {cost}")
        else:
            print("Low balance")

    def check_balance(self):
        print(f"Owner: {self.owner_name} | Balance: Rs. {self.balance}")

# Test
sim = MobileSIM("03001234567", "Usman", 10.0)
sim.check_balance()
sim.make_call(6)  # Cost 12 > Balance 10 (Fails)
sim.recharge(50)
sim.make_call(6)  # Cost 12 (Succeeds)
sim.check_balance()
```

```python
#Task 11: Parking Fee
class ParkingTicket:
    def __init__(self, vehicle_number, hours_parked):
        self.vehicle_number = vehicle_number
        self.hours_parked = hours_parked

    def calculate_fee(self):
        if self.hours_parked <= 2:
            return 50
        else:
            extra_hours = self.hours_parked - 2
            return 50 + (extra_hours * 30)

    def print_ticket(self):
        fee = self.calculate_fee()
        print(f"Vehicle No: {self.vehicle_number} | Hours: {self.hours_parked} | Total Fee: Rs. {fee}")

# Test for 1, 2, and 5 hours
t1 = ParkingTicket("LEB-1234", 1)
t2 = ParkingTicket("LEA-5678", 2)
t3 = ParkingTicket("LEC-9999", 5)

t1.print_ticket()
t2.print_ticket()
t3.print_ticket()
```

```python
#Task 12: Electricity Bill
class ElectricityBill:
    def __init__(self, name, meter_number, previous_reading, current_reading):
        self.name = name
        self.meter_number = meter_number
        self.previous_reading = previous_reading
        self.current_reading = current_reading

    def calculate_units(self):
        return self.current_reading - self.previous_reading

    def calculate_bill(self):
        units = self.calculate_units()
        bill = 0

        if units <= 100:
            bill = units * 10
        elif units <= 200:
            bill = (100 * 10) + ((units - 100) * 15)
        else:
            bill = (100 * 10) + (100 * 15) + ((units - 200) * 20)

        return bill

    def display_bill(self):
        units = self.calculate_units()
        total_bill = self.calculate_bill()
        print(f"Customer: {self.name} | Meter #: {self.meter_number}")
        print(f"Units Consumed: {units} | Total Bill: Rs. {total_bill}\n" + "-"*40)

# Test with 2 customers
c1 = ElectricityBill("Hamza", "M-101", 1000, 1150)  # 150 units
c2 = ElectricityBill("Zainab", "M-102", 2000, 2250) # 250 units

c1.display_bill()
c2.display_bill()
```

```python
#Task 13: Quiz Competition
class Participant:
    def __init__(self, name, total_questions, correct_answers):
        self.name = name
        self.total_questions = total_questions
        self.correct_answers = correct_answers

    def calculate_percentage(self):
        return (self.correct_answers / self.total_questions) * 100

    def find_grade(self):
        pct = self.calculate_percentage()
        if pct >= 80:
            return 'A'
        elif pct >= 60:
            return 'B'
        elif pct >= 40:
            return 'C'
        else:
            return 'F'

    def display(self):
        print(f"Participant: {self.name}")
        print(f"Percentage: {self.calculate_percentage():.2f}% | Grade: {self.find_grade()}\n")

# Test with 3 participants
p1 = Participant("Bilal", 20, 18)   # 90% -> Grade A
p2 = Participant("Ayesha", 20, 13)  # 65% -> Grade B
p3 = Participant("Tariq", 20, 7)    # 35% -> Grade F

participants = [p1, p2, p3]

# Display all results
for p in participants:
    p.display()

# Find and print the winner (highest percentage)
winner = max(participants, key=lambda p: p.calculate_percentage())
print(f"Winner with the highest score: {winner.name} ({winner.calculate_percentage():.2f}%)")
```
